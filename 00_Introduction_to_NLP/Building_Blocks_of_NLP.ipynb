#### Import necessary libaries

```python
import re # used for regular expressions
from collections import Counter # used to count frequency of items in an iterable (list, string, etc.)
```

### 1. Corpus - a collection of text data for an NLP task
In this case, it will be the quotes that I asked you to submit at the beginning of the semester.


### 2. Document - **one** individual unit of text inside a corpus
In this case, it is one particular quote that is counted as a document.

```python
# Since we already loaded it a while ago
# However, it is currently loaded as one big string

# I added a '---' delimiter to mark the end of a quote
# We split there to create the corpus
# Splitting the text returns a list
splitted_quotes = quotes.split("---")

# Populate the corpus
corpus = []
for quote in splitted_quotes:
    quote = quote.strip()
    corpus.append(quote)

# Check the count of documents in the corpus and show the first two quotes
print(f"Total Quotes (Documents): {len(corpus)}\n")
print(f"Document 1: {corpus[0]}\n")
print(f"Document 2: {corpus[1]}")
```

### 3. Token - a single "piece" of text after splitting
It is normally what we consider as a word, but it can depend on how granular we will be. We can include punctuation marks (`!`, `.`, `;`) or not. Again, it depends.

### 4. Tokenization - the process of breaking a document into tokens
It can be done by regular expressions (regex) or a dedicated tokenizer available in Python libraries.

### 5. Word - a token that is specifically a word (not punctuation, not a number)
Just to make a distinction from word and token.

```python
def tokenize(document):
    # Pre-processing text by setting everything to lowercase
    document = document.lower()

    # \w+ considers letters and numbers that are continuously connected (Example: abc123)
    tokens = re.findall(r"\w+", document)

    return tokens
```

```python
tokenized_corpus = []

# Loop through each document in the corpus
for document in corpus:
    tokenized_corpus.append(tokenize(document))

print(tokenized_corpus)
```

```python
# View all the quotes and their tokenized versions
for i, (doc, tokens) in enumerate(zip(corpus, tokenized_corpus)):
    print(f"Document {i+1}: {doc}\n")
    print(f"Tokens ({len(tokens)}): {tokens}\n")
```

### 6. Vocabulary - the set of unique tokens or words seen across the whole corpus.

```python
# Compile all the tokens from the documents
all_tokens = []
for tokens in tokenized_corpus:
    for token in tokens:
        all_tokens.append(token)

# Keep only the unique tokens, the set() automatically handles duplicates for us
unique_tokens = set()
for token in all_tokens:
    unique_tokens.add(token)

# Sort them alphabetically so the output is easy to read
vocabulary = sorted(unique_tokens)

print(f"Total tokens (with repeats): {len(all_tokens)}")
print(f"Vocabulary size: {len(vocabulary)}")
print(f"Vocabulary: {vocabulary}\n")
```

### Small exercise.
1. What is the longest document? How about the shortest document?
2. What is the average token length of a document in the corpus?
3. What is the longest token? How about the shortest token?
4. What are the top 5 most frequent tokens? What are their counts? Use the Counter library and the `most_common` function of that library.
5. Are the tokens meaningful by themselves?

```python
# 1) Longest document and shortest document

# Indices for longest and shortest documents
longest_idx = 0
shortest_idx = 0

# Loop through all the tokenized corpus
for i in range(len(tokenized_corpus)):
    # Search for the longest document
    if len(tokenized_corpus[i]) > len(tokenized_corpus[longest_idx]):
        longest_idx = i
    # Search for the shortest document
    if len(tokenized_corpus[i]) < len(tokenized_corpus[shortest_idx]):
        shortest_idx = i

# Get the documents and number of tokens using the indices
longest_doc = corpus[longest_idx]
longest_tokens = tokenized_corpus[longest_idx]
shortest_doc = corpus[shortest_idx]
shortest_tokens = tokenized_corpus[shortest_idx]

print(f"Longest Document: {longest_doc}\n")
print(f"Numberer of Tokens: {len(longest_tokens)}\n")
print("===" * 10)
print()
print(f"Shortest Document: {shortest_doc}\n")
print(f"Number of Tokens: {len(shortest_tokens)}")
```

```python
# 2) Average token length of a document in the corpus
total_tokens = 0
total_docs = 0

# Loop through the tokenized_corpus to calculate totals
for document_tokens in tokenized_corpus:
    # Count the number of tokens in this document
    total_tokens += len(document_tokens)
    # Count the number of total documents
    total_docs += 1

# Calculate the average
if total_docs > 0:
    avg_doc_length = total_tokens / total_docs
else:
    avg_doc_length = 9

print(f"Total Documents: {total_docs}")
print(f"Total Tokens across Corpus: {total_tokens}")
print(f"Average Document Word Length: {avg_doc_length:.2f} token(s) per document")
```

```python
# 3) Longest token and shortest token

longest_token = ""
shortest_token = None

# Iterate through every document and every token
for tokens in tokenized_corpus:
    for token in tokens:
        # Set initial value for shortest_token on the first iteration
        if shortest_token is None:
            shortest_token = token

        # Compare string character lengths
        if len(token) > len(longest_token):
            longest_token = token

        if len(token) < len(shortest_token):
            shortest_token = token

print(f"Longest Token: '{longest_token}' (Length: {len(longest_token)})")
print(f"Shortest Token: '{shortest_token}' (Length: {len(shortest_token)})")
```

```python
# 4) Top 5 most frequent tokens and their counts

# Counter automatically counts the instances for us
word_freq = Counter(all_tokens)

print("Top 5 most common tokens across the corpus:")
# Most common does the sorting based on the counts
for word, count in word_freq.most_common(5):
    # !r wraps the word in quotation marks
    print(f"  {word!r}: {count}")
```
