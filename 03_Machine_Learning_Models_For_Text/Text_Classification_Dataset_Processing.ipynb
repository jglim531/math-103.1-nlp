### Load Dataset of Airline Sentiment

Link: https://huggingface.co/datasets/osanseviero/twitter-airline-sentiment

A sentiment analysis job about the problems of each major U.S. airline. Twitter data was scraped from February of 2015 and contributors were asked to first classify positive, negative, and neutral tweets, followed by categorizing negative reasons (such as "late flight" or "rude service").

Generally, the dataset contains whether the sentiment of the tweets was positive, neutral, or negative for six US airlines.

```python
# How to load dataset from Hugging Face using the dataset title
airline_dataset = load_dataset("osanseviero/twitter-airline-sentiment")
```

```python
# See the details of the dataset (this one only contains the train split, so will need to further split this data for train and test sets)
airline_dataset
```

```python
# Get the train dataset from the DatasetDict
airline_train = airline_dataset["train"]

# Convert it to .csv for a local copy
airline_train.to_csv("airline_sentiment.csv")
```

```python
# Convert to pandas directly after loading from Hugging Face
airline_dataset = airline_dataset["train"].to_pandas()

# Show the top N entries
airline_dataset.head()
```

```python
# Inspect entries of the train dataset
print("Airline Dataset (Train) - Two Examples per Label\n")

for sentiment in ["negative", "neutral", "positive"]:
    # Filter the subset to a specific sentiment then select two samples, random state makes it reproducible per run
    subset = airline_dataset[airline_dataset["airline_sentiment"] == sentiment].sample(2, random_state = 42)
    for _, row in subset.iterrows():
        print(f"Label: {row['airline_sentiment']}")
        print(f"Text:  {row['text']}\n")
```

```python
# Load from local copy
airline_test = pd.read_csv("/content/airline_sentiment.csv")

airline_test.head()
```

### Load Data of Financial News Sentiment

Link: https://huggingface.co/datasets/zeroshot/twitter-financial-news-sentiment

The Twitter Financial News dataset is an English-language dataset containing an annotated corpus of finance-related tweets. This dataset is used to classify finance-related tweets for their sentiment.

Generally, the dataset contains whether the sentiment of the tweets (related to the stock market) was bearish, bullish, or neutral for six US airlines.

```python
# Load data from Hugging Face
financial_dataset = load_dataset("zeroshot/twitter-financial-news-sentiment")
```

```python
# Check downloaded datasets
financial_dataset
```

```python
# The link to the dataset in Hugging Face defines the label as follows
id_to_label = {
    0: "Bearish",
    1: "Bullish",
    2: "Neutral"
}
```

```python
# Add a new column called text_label to both train and validation datasets
financial_train = financial_dataset["train"].to_pandas() # Need to convert the Hugging Face dataset to pandas DataFrame first
financial_train["text_label"] = financial_train["label"].apply(lambda x: id_to_label[x])

financial_validation = financial_dataset["validation"].to_pandas() # Need to convert the Hugging Face dataset to pandas DataFrame first
financial_validation["text_label"] = financial_validation["label"].apply(lambda x: id_to_label[x])
```

```python
# Convert it to .csv for a local copy
financial_train.to_csv("financial_sentiment_train.csv")
financial_validation.to_csv("financial_sentiment_validation.csv")
```

```python
# Inspect entries of the train dataset
print("Financial Dataset (Train) - Two Examples per Label")
financial_train.groupby("label").head(2).sort_values("label")
```

```python
print("Two Example Entries per Label of Financial Dataset (Train)\n")

for label_id in id_to_label:
    subset = financial_train[financial_train["label"] == label_id].sample(2, random_state = 42)
    for _, row in subset.iterrows():
        print(f"Label: {row['label']} ({row['text_label']})")
        print(f"Text:  {row['text']}\n")
```

```python
# Inspect entries of the validation dataset
print("Financial Dataset (Validation) - Two Examples per Label")
financial_validation.groupby("label").head(2).sort_values("label")
```
