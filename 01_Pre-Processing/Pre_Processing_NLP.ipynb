### 1. Sentence Segmentation

````python
import spacy
nlp = spacy.load("en_core_web_sm")

text = ("The only way to do great work is to love what you do. "
        "I think, therefore I am. "
        "That's one small step for man, one giant leap for mankind.")

# Automatically handles sentence segmentation, tokenization, and other NLP tasks
doc = nlp(text)

# Access the sentences in the processed document
for sent in doc.sents:
    print(f"Sentence: {sent.text}")
````

### 2. Word Tokenization

````python
doc = nlp("Whatever you are, be a good one.")

tokens = []

# Access the tokens in the processed document
# The tokens are stored in the `doc` object and has a number of attributes and methods that can be used to access the tokens and their properties
for token in doc:
    tokens.append(token.text)

print(tokens)
# ['Whatever', 'you', 'are', ',', 'be', 'a', 'good', 'one', '.']
````

### 3. Stemming
Done by NLTK since spaCy defaults to lemmatization

````python
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

# Create a PorterStemmer object
stemmer = PorterStemmer()

# Tokenize the text into words and apply stemming
tokens = word_tokenize("The future belongs to those who believe in the beauty of their dreams.")

stems = []

for token in tokens:
    stemmed_token = stemmer.stem(token)
    print(f"Token: {token}, Stem: {stemmed_token}")
    stems.append(stemmed_token)

print(stems)
# ['the', 'futur', 'belong', 'to', 'those', 'who', 'believ', 'in', 'the', 'beauti', 'of', 'their', 'dream', '.']
````

### 4. Lemmatization

````python
doc = nlp("The future belongs to those who believe in the beauty of their dreams.")

lemmas = []

for token in doc:
    # Access the lemma of each token using the `lemma_` attribute
    lemma = token.lemma_
    print(f"Token: {token.text}, Lemma: {lemma}")
    lemmas.append(lemma)

print(lemmas)
# ['the', 'future', 'belong', 'to', 'those', 'who', 'believe', 'in', 'the', 'beauty', 'of', 'their', 'dream', '.']
````

### 5. Stopword Removal

````python
doc = nlp("The only thing we have to fear is fear itself.")

filtered = []

for token in doc:
    # Check if the token is not a stopword using the `is_stop` attribute
    if not token.is_stop:
        filtered.append(token.text)

print(filtered)
# ['thing', 'fear', 'fear', '.']
````

### 6. Special Character Filtering

````python
doc = nlp("Elementary, my dear Watson.")

# Remove punctuation and non-alphabetic tokens
clean = []

for token in doc:
    # Check if the token is alphabetic using the `is_alpha` attribute
    if token.is_alpha:
        clean.append(token.text)

print(clean)
# ['Elementary', 'my', 'dear', 'Watson']
````

### 7. Casing Standardization

````python
# Does not need any library but is a pre-processing step in NLP tasks
text = "Be the change that you wish to see in the world."

print(text.lower())
# "be the change that you wish to see in the world."
````

### 8. Part-of-Speech (POS) Tagging

````python
doc = nlp("A journey of a thousand miles begins with a single step.")

# Uses .pos_ and .tag_ attributes to access the part-of-speech (POS) and detailed POS tag of each token
for token in doc:
    print(f"Text: {token.text}, POS: {token.pos_}, Tag: {token.tag_}")
# journey  NOUN  NN
# begins   VERB  VBZ
# single   ADJ   JJ
# step     NOUN  NN
# ...
````

### 9. Named Entity Recognition (NER)

````python
doc = nlp("Neil Armstrong said this in 1969, and John F. Kennedy spoke to Americans a few years earlier.")

# Named entities in doc
for ent in doc.ents:
    # .text - the actual text span of the entity
    # .label_ - the category spaCy assigned to it
    print(f"Text: {ent.text}, Label: {ent.label_}")
# Neil Armstrong, PERSON
# 1969, DATE
# John F. Kennedy, PERSON
# a few years earlier, DATE
````

### Load the dataset

````python
import re

file_path = "updated_quotes.txt"

# Open and read the file
with open(file_path, 'r', encoding='utf-8') as file:
    quotes = file.read()
````

Suggested pre-processing steps:
1. Split the text into entries on lines containing only --- (three or more dashes on their own line)
2. Within each entry collapse consecutive blank lines ("double enters") down to a single newline
3. Strip leading/trailing whitespace from each entry

````python
# Split on lines that are just dashes (---)
entries = re.split(r'\n-{3,}\n', quotes)

cleaned_entries = []

for entry in entries:
    # Collapse double (or more) newlines into a single newline
    entry = re.sub(r'\n\s*\n+', '\n', entry)
    # Strip leading/trailing whitespace
    entry = entry.strip()
    if entry:
        cleaned_entries.append(entry)

print(f"Total of {len(cleaned_entries)} entries\n")

for x in range(len(cleaned_entries)):
    print(f"Quote {x + 1}: {cleaned_entries[x]}\n")
````
